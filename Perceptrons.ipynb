```python
import numpy as np

# Step Activation Function
def step_function(x):
    if x >= 0:
        return 1
    else:
        return 0

# Training Data (AND Gate)
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Target Output
y = np.array([0, 0, 0, 1])

# Initialize weights and bias
weights = np.zeros(2)
bias = 0

# Learning rate
learning_rate = 0.1

# Number of epochs
epochs = 10

# Training
for epoch in range(epochs):
    print(f"\nEpoch {epoch + 1}")

    for i in range(len(X)):
        # Calculate weighted sum
        net_input = np.dot(X[i], weights) + bias

        # Apply activation function
        prediction = step_function(net_input)

        # Calculate error
        error = y[i] - prediction

        # Update weights and bias
        weights += learning_rate * error * X[i]
        bias += learning_rate * error

        print(f"Input: {X[i]}, Target: {y[i]}, Prediction: {prediction}, Error: {error}")

print("\nTraining Completed!")
print("Final Weights:", weights)
print("Final Bias:", bias)

# Testing
print("\nTesting Perceptron")
for i in range(len(X)):
    net_input = np.dot(X[i], weights) + bias
    prediction = step_function(net_input)
    print(f"Input: {X[i]} --> Output: {prediction}")
```

```python
import numpy as np

# Sigmoid Activation Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Derivative of Sigmoid
def sigmoid_derivative(x):
    return x * (1 - x)

# XOR Dataset
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])

y = np.array([[0],
              [1],
              [1],
              [0]])

# Seed for reproducibility
np.random.seed(42)

# Initialize weights and biases
input_neurons = 2
hidden_neurons = 4
output_neurons = 1

W1 = np.random.uniform(-1, 1, (input_neurons, hidden_neurons))
b1 = np.random.uniform(-1, 1, (1, hidden_neurons))

W2 = np.random.uniform(-1, 1, (hidden_neurons, output_neurons))
b2 = np.random.uniform(-1, 1, (1, output_neurons))

learning_rate = 0.5
epochs = 10000

# Training
for epoch in range(epochs):

    # ---------- Forward Propagation ----------
    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2) + b2
    predicted_output = sigmoid(final_input)

    # ---------- Error ----------
    error = y - predicted_output

    # ---------- Backpropagation ----------
    d_output = error * sigmoid_derivative(predicted_output)

    hidden_error = np.dot(d_output, W2.T)
    d_hidden = hidden_error * sigmoid_derivative(hidden_output)

    # ---------- Update Weights ----------
    W2 += learning_rate * np.dot(hidden_output.T, d_output)
    b2 += learning_rate * np.sum(d_output, axis=0, keepdims=True)

    W1 += learning_rate * np.dot(X.T, d_hidden)
    b1 += learning_rate * np.sum(d_hidden, axis=0, keepdims=True)

    # Print loss every 1000 epochs
    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss = {loss:.6f}")

# ---------- Testing ----------
print("\nFinal Predictions")

for i in range(len(X)):
    hidden = sigmoid(np.dot(X[i], W1) + b1)
    output = sigmoid(np.dot(hidden, W2) + b2)
    print(f"Input: {X[i]} --> {output[0][0]:.4f}")
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# XOR Dataset
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Create MLP
model = Sequential()

# Hidden Layer
model.add(Dense(4, activation='relu', input_shape=(2,)))

# Output Layer
model.add(Dense(1, activation='sigmoid'))

# Compile
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train
model.fit(X, y,
          epochs=1000,
          verbose=0)

# Predictions
predictions = model.predict(X)

print("Predictions:")
print(predictions.round())
```
