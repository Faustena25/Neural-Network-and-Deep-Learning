Load the dataset

Removeing Fake values

```python
cols_to_check = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

for col in cols_to_check:
    zero_count = (df[col] == 0).sum()
    print(f"{col:30} has {zero_count} zeros")
```

```python
# Replace fake zeros with the median value of each column
cols_to_fix = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

for col in cols_to_fix:
    median_value = df[col].median()
    df[col] = df[col].replace(0, median_value)

```

```python
for col in cols_to_fix:
    zero_count = (df[col] == 0).sum()
    print(f"{col:30} zeros remaining: {zero_count}")
```

 Normalize the Features

```python
X = df.drop('Outcome', axis=1).values
y = df['Outcome'].values

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print("Before normalization:")
print(f"  Age range:     {X[:,7].min():.0f}  to  {X[:,7].max():.0f}")
print(f"  Insulin range: {X[:,4].min():.0f}  to  {X[:,4].max():.0f}")
```

```python
# Normalize using StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python

print("After normalization:")
print(f"  Age mean:     {X_scaled[:,7].mean():.4f}  ")
print(f"  Insulin mean: {X_scaled[:,4].mean():.4f}  ")
print(f"  Age std:      {X_scaled[:,7].std():.4f}  ")
print(f"  Insulin std:  {X_scaled[:,4].std():.4f}   ")
```

 Split into Training and Testing Sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y,
    test_size=0.2,
    random_state=42,
)

print("Training samples:", X_train.shape[0])
print("Testing samples: ", X_test.shape[0])
```

```python
# Check class balance in both sets
train_diabetes = y_train.sum()
test_diabetes  = y_test.sum()

print("Training set:")
print(f"  Healthy (0): {len(y_train) - train_diabetes}  |  Diabetic (1): {train_diabetes}")

print("\nTesting set:")
print(f"  Healthy (0): {len(y_test) - test_diabetes}  |  Diabetic (1): {test_diabetes}")
```
